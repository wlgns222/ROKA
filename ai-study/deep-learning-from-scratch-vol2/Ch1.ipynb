### 1.1.1 벡터와 행렬

**벡터** : 크기와 방향을 가진 양. 숫자가 일렬로 늘어선 형태로 표현가능하며, 파이썬에서는 1차원 배열로 취급할 수 있습니다.

**행렬** : 숫자가 2차원 형태로 늘어선 것이다. 파이썬에서는 2차원 배열로 표현할 수 있다. 가로줄을 행, 세로줄을 열이라고 한다.

파이썬에서 벡터와 행렬은 np.array 에서 생성할 수 있다.

```python
import numpy as np

x = np.array([1, 2, 3])
print(x.shape) # 배열의 형상
print(x.ndim) # 차원 수

W = np.array ([[1,2,3], [4,5,6]])
print(W.shape) # 배열의 형상
print(W.ndim) # 차원 수
```

### 1.1.2 행렬의 원소별 계산

```python
X = np.array ([[0,1,2], [3,4,5]])

print("원소별 합 \n", W + X)
print("원소별 곱 \n", W * X)
```

### 1.1.3 브로드캐스트

```python
A = np.array([[1,2], [3,4]])
print(A*10)
```

위 예시에서 스칼라 값 10 은 2x2 행렬로 확장된 후 원소별 연산을 수행한다.

이 영리한 기능을 **브로드캐스트** 라고 한다.
