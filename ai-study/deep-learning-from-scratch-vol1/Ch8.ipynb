### 8.1.3 깊게 하는 이유

층 쌓기의 중요성은 ILSVRC 로 대표되는 대규모 이미지 인식 대회의 결과에서 파악할 수 있다. 해당 대회에서 상위를 차지한 대부분은 딥러닝 기반이며, 신경망을 더 깊게 만드는 방향으로 가고 있다.

**깊은 신경망의 장점**

**1. 신경망의 매개변수 수가 줄어든다.**

층을 깊게 한 신경망은 깊지 않은 경우보다 적은 매개변수로 같은 (혹은 그 이상) 수준의 표현력을 달성할 수 있다.

예를 들어, 5x5 합성곱 연산 1회는 3x3 합성곱 연산 2회 수행하여 대체할 수 있다.

이때, 전자는 25개의 매개변수가 필요하지만 후자는 18개의 매개변수가 필요하다.

층이 깊어질 수록 해당 차이는 심해진다.

7x7 합성곱 연산은 3x3 합성곱 연산 3회로 수행할 수 있다.

이때, 전자는 49개의 매개변수가 후자는 27개의 매개변수가 필요하다.

**2. 학습의 효율성**

층을 깊게 함으로써 학습 데이터의 양을 줄여 학습을 고속으로 수행할 수 있다.

예를 들어 '개'를 인식하는 문제에서 얕은 신경망에서 해결하기 위해서는 합성곱 계층은 개의 특징 대부분을 이해하여야 한다.

그러나 신경망을 깊게하면 학습해야할 문제를 계층적으로 분해할 수 있다.

처음층은 에지학습에 전념하여 학습 데이터를 효율적으로 활용할 수 있고, 해당 정보를 계층적으로 전달하여 다음 층은 그 정보를 쓸 수 있다.

## 8.2 딥러닝의 초기 역사

딥러닝이 큰 주목을 받게 된 계기는 이미지 인식 기술을 겨루는 장인 ILSVRC 의 2012년 대회다.

해당 대회에서 딥러닝에 기초한 기법, 일명 AlexNet이 압도적인 성적으로 우승하면서 이미지 인식에 대한 접근법을 뒤흔들었다.

### 8.2.1 이미지 넷

**ImageNet** 은 100만 장이 넘는 이미지를 담고 있는 데이터 셋이다. 2017년까지 매년 열린 ILSVRC 는 해당 데이터셋을 사용하여 자웅을 겨루는 대회이다.

2012년 AlexNet이 오류율을 크게 낮췄고, 2015년에는 150층이 넘는 심층 신경망인 ResNet 이 오류율을 3.5% 까지 낮췄다.

마지막 대회인 2017년의 우승모델인 SENets 는 2.3% 까지 끌어내렸다.

그 중 VGG, GoogleLeNet, ResNet은 특히 유명하다.

### 8.2.2 VGG

**VGG** 는 합성곱 계층과 풀링 계층으로 구성되는 '기본적인 CNN'이다.

비중있는 층(합성곱 계층, 완전연결 계층)을 모두 16층(혹은 19층)으로 심화한게 특징이다.

3x3 의 필터를 사용한 합성곱 계층을 연속으로 거친다. 합성곱 연산을 3~4호히 연속으로 풀링 계층을 두어 크기를 절반으로 줄이는 처리를 반복한다.

### 8.2.3 GoogleLeNet

기본적으로 지금까지 보아온 CNN과 크게 다르지 않다.

**다만, GoogleLeNet은 세로 방향 깊이 뿐 아니라 가로 방향도 깊다.**

GoogleLeNet 에는 가로방향 '폭'이 있다. 이를 인셉션 구조라 한다.
